```python
# same function is performing different actions for different inputs
def add(a,b): 
    return a+b
print(add(3,2))# addition
print(add('code', 'gnan'))# concatenation
```

```python
# method overriding
# person
class Person():
    def __init__(self,name,age):
        self.name=name
        self.age=age    
    def show(self):
       return  f'My self {self.name} and i am a {self.age} years old'
# student
class Student(Person):
    def __init__(self,name,age,std):
        # super().__init__(name,age)
        Person.__init__(self,name,age)
        self.std=std    
    def show(self):
       return f'My self {self.name} and i am studing {self.std} class'
p=Person('karthik',20)
s=Student('karhikeya',15,10)
print(p.show())
print(s.show())
```

```python
# method overriding
# person
class Person():
    def __init__(self,name,age):
        self.name=name
        self.age=age    
    def show(self):
       return  f'My self {self.name} and i am a {self.age} years old'
# student
class Student(Person):
    def __init__(self,name,age,std):
        # super().__init__(name,age)
        Person.__init__(self,name,age)
        self.std=std    
    def show(self):
        print(super().show())
        return f"My self {self.name} and i am studing {self.std} class"
p=Person('karthik',20)
s=Student('karhikeya',15,10)
# print(p.show())
print(s.show())
```

```python
# method overloading 
class Math:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):
        return a+b+c
    def add(self,a,b,c,d):
        return a+b+c+d
    
m=Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
# method overloading  by using default arguments
class Math: 
    def add(self,a,b,c = 0,d = 0):
        return a+b+c+d   
m=Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
# method overloading  by using default arguments
class Math: 
    def add(self,a,b,c = 0,d = 0, *args):
        print(type(args))
        return a+b+c+d   
m=Math()# *args = 5 6 7 8
print(m.add(1,2,3,4,5,6,7,8))
print(m.add(1,2,3))
print(m.add(1,2))
```
