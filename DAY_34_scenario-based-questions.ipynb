Design a Python class BankAccount that:
Stores account holder name, account number, and balance
Implements the following methods:
      deposit(amount)
      withdraw(amount)
      check_balance()
Ensures balance cannot become negative.

```python
class BankAccount:
    def __init__(self, name, account_number, balance):
        self.name = name
        self.account_number = account_number
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance -= amount
            print("Withdrawn:", amount)
    def check_balance(self):
        print("Current Balance:", self.balance)
# Object creation
acc1 = BankAccount("Karthikeya", 12345, 5000)
acc1.deposit(2000)
acc1.withdraw(3000)
acc1.check_balance()
```

Create a Python OOP design where:
A base class Product contains:
    name
    price
method display_details()
Two subclasses:
    Electronics → includes warranty
    Clothing → includes size and material
Override the display_details() method in each subclass.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def display_details(self):
        print("Product:", self.name)
        print("Price:", self.price)
class Electronics(Product):
    def __init__(self, name, price, warranty):
        super().__init__(name, price)
        self.warranty = warranty
    def display_details(self):
        print("Electronics:", self.name)
        print("Price:", self.price)
        print("Warranty:", self.warranty)
class Clothing(Product):
    def __init__(self, name, price, size, material):
        super().__init__(name, price)
        self.size = size
        self.material = material
    def display_details(self):
        print("Clothing:", self.name)
        print("Price:", self.price)
        print("Size:", self.size)
        print("Material:", self.material)
e1 = Electronics("Laptop", 50000, "2 Years")
c1 = Clothing("T-Shirt", 1000, "M", "Cotton")
e1.display_details()
print()
c1.display_details()
```
